## 2. The question: decision, action, cost of a wrong call

Search question: Which existing pages should a content team review first
because they may be losing search visibility?

Unit of analysis: one page (one row per page).

Output: a ranked review list, most at-risk pages first, with reason codes
explaining why each page ranks high.

Action: a content team reviews the top-ranked pages and decides whether to
refresh, expand, or monitor each one.

Cost of a wrong recommendation: if a healthy page is ranked high, the team
wastes limited review time. If a page that really needed attention is ranked
low, it keeps losing visibility unnoticed. A recommendation is a prompt for
human review, not a guarantee that editing the page will help.

Why data or ML can help: there are too many pages to review by hand, and in
Notebook 01 a simple hand-written rule found only about 24% correct in its
top 50, while a learned model scored higher. The value is in the decision it
improves (who gets reviewed first), not in the algorithm itself.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# The small starter dataset that ships with the repo
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Same filter the pipeline uses: pages with visibility and enough history
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]

# Label used in the starter pipeline (a proxy for decline)
df["is_declining_label"] = (df["trend_direction"] == "down")

# Number 1: how many pages
print(f"Pages in the starter dataset: {len(df):,}")

# Number 2: how common decline is
share_down = df["is_declining_label"].mean()
print(f"Share of pages trending down: {share_down:.1%}")

# Number 3: is there a stale but visible group worth reviewing?
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
print(f"Stale (180+ days) pages with 500+ impressions: {stale_visible:,}")
```

Observed: of 30,000 pages, 54.2% were trending down, so roughly half of the
inventory shows a decline signal, which is far too many pages to review by
hand. Only 17 pages were both stale (180+ days since update) and visible
(500+ impressions), which suggests a fixed rule like this catches a very
small group, while a learned ranking can prioritize across the whole set.
This is an observation on the starter dataset, not a claim about all clients.

Note: the "declining" label is a proxy based on the current window's
trend_direction, so 54.2% describes how the label is defined here, not
the true rate of decline.
