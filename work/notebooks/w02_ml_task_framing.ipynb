## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Same filter as the pipeline: pages with visibility and enough history
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].copy()

# The proxy target
df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

print("Unit of analysis: one row = one page")
print(f"Rows: {len(df):,}, Columns: {df.shape[1]}")

# Show a real slice of the data: id, a few features, and the target
cols_to_show = [
    "content_id", "impressions_90d", "avg_position", "ctr",
    "content_age_days", "trend_direction", "is_declining_label",
]
df[cols_to_show].head(10)
```

## 5. Why ML beats a fixed rule here

Why ML, not just a rule: A hand-written rule like "flag pages with 180+
days since update and 500+ impressions" only found 17 pages in the starter
dataset, and scored 0.240 Precision@50 in Notebook 01/02. A learned model
scored much higher (0.680 to 0.820 at different depths, in-sample), because
it can combine many signals (impressions, position, CTR, age, freshness)
and weigh them automatically, instead of relying on a few fixed thresholds
a person guessed. With 30,000+ pages, no one can manually inspect every
combination of signals, so a model that ranks by combined evidence adds
real value over a simple if/else rule.
